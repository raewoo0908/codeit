### 문제1. 3의 배수 출력

- **문제 설명**

- **문제 설명**  
문자열에서 모음(a, e, i, o, u)을 제거한 문자열을 반환하는 함수를 작성하세요.

- **함수 설명**  
`remove_vowels(s: string) -> string`:  
  - s: 영어 소문자 알파벳으로 이루어진 문자열입니다.

- **입출력 예시**   

    - 입력1:

    ```
    "codeit"
    ```

    - 출력1:

    ```
    "cdt"
    ```

    - 입력2:
    
    ```
    "python programming"
    ```

    - 출력2:

    ```
    "pythn prgrmmng"
    ```



````python
# 문제
def remove_vowels(s):
    vowels = "aeiouAEIOU"
    result = ""
    for char in s:
        if char not in vowels:
            result += char
    return result
````

````python
# 프로그램 실행
print(remove_vowels("codeit"))
print(remove_vowels("python_programming"))
````

- **입출력 예시**

- **함수 설명**

  `sum_even_numbers(n: int)`

  - 사용자에게 숫자 입력값을 받아, 1부터 n까지의 짝수 합은 ___ 입니다." 가 출력됩니다.

- **입출력 예시**

    - 입력1:

    - 출력2:


````python
# 문제
def sum_even_numbers(n):
  if n == 0: 
    print("종료")
    return 
  sum = 0
  for i in range(2, n + 1, 2):
    sum += i
  print(f"\"1부터 {n}까지의 짝수의 합은 {sum}입니다.\"")
  return
````

### 문제1. 문자열 압축 게임



- **문제 설명**

  주어진 문자열에서 연속해서 같은 문자가 반복되는 경우, 그 문자와 반복 횟수를 이용해 새로운 형태로 문자열을 압축합니다. 예를 들어, "aabcccccaaa" 문자열은 "a2b1c5a3"로 압축될 수 있습니다.

- **함수 설명**  
`compress_string(s: str) -> str`:  
  - s: 압축하고자 하는 문자열입니다.

- **입출력 예시**

    - 입력1:

    ```
    "aaabbaaa"
    ```

    - 출력1:

    ```
    "a3b2a3"
    ```

    - 입력2:
    
    ```
    "aabcccccaaa"
    ```

    - 출력2:

    ```
    "a2b1c5a3"
    ```

````python
# 문제
from itertools import groupby

def compress_string(s):

    result = ""
    for k, g in groupby(s):  # TODO: groupby 용례 더 공부하기
        result += k + str(len(list(g)))

    return result
````

````python
# 프로그램 실행
print(compress_string("aaabbaaa"))
print(compress_string("aabcccccaaa"))
````

### 문제2. 덧셈 프로그램



- **문제 설명**
이 프로그램은 두 개의 숫자를 입력받아 그 합을 계산합니다. 입력받은 값이 숫자 형식이 아니면, 사용자에게 숫자 형식의 입력을 요청합니다. 숫자 입력이 완료되면 두 숫자의 합을 출력합니다.

- **함수 설명**

  `add_numbers()`
  - 이 함수는 `input()`을 사용하여 사용자로부터 두 개의 숫자를 입력받습니다.
  - 입력된 값이 숫자가 아닐 경우, "올바른 숫자를 입력하세요."라는 메시지를 출력하고 다시 입력을 요청합니다.
  - 두 숫자의 합을 계산하고 결과를 출력합니다.

- **입출력 예시**

    - 입력:
        ```
        첫 번째 숫자를 입력하세요: 10
        첫 번째 숫자를 입력하세요: twenty
        첫 번째 숫자를 입력하세요: 10
        두 번째 숫자를 입력하세요: 20
        ```

    - 출력:
        ```
        올바른 숫자를 입력하세요.
        10 + 20 = 30
        ```

````python
int(input()) # Value Error
````

````python
# 문제
def add_numbers():

    while True:
        input1 = input("첫 번째 숫자를 입력하세요: ")

        try:
            num1 = int(input1)
        except ValueError:
            print("올바른 숫자를 입력하세요.")
            continue
        break

    while True:
            input2 = input("두 번째 숫자를 입력하세요: ")
    
            try:
                num2 = int(input2)
            except ValueError:
                print("올바른 숫자를 입력하세요.")
                continue
            break

    print(f"{num1} + {num2} = {num1+num2}")
````

````python
# 프로그램 실행
import builtins

def run_example(title, inputs):
    """미리 준비한 입력값을 순서대로 넣어 add_numbers()를 실행합니다."""
    print(f"[{title}] 입력값: {inputs}")

    values = iter(inputs)
    real_input = builtins.input

    def fake_input(prompt=""):
        value = next(values)
        print(f"{prompt}{value}")   # 사용자가 직접 입력한 것처럼 화면에 보여주기
        return value

    builtins.input = fake_input          # input()을 준비한 값으로 대체
    try:
        add_numbers()
    finally:
        builtins.input = real_input      # 예시가 끝나면 원래 input()으로 되돌리기
    print()


# 1) 에러 없음: 두 번 모두 정상적인 숫자를 입력한 경우
run_example("정상 입력", ["10", "20"])

# 2) 첫 번째 숫자에 문자를 입력한 경우 -> ValueError -> "올바른 숫자를 입력하세요." 후 재입력
run_example("첫 번째 숫자에 문자 입력", ["twenty", "10", "20"])

# 3) 두 번째 숫자에 문자를 입력한 경우 -> ValueError -> 두 번째 입력만 다시 요청
run_example("두 번째 숫자에 문자 입력", ["10", "twenty", "20"])

# 4) 실수/빈 값/공백/기호/쉼표 등 int()로 바꿀 수 없는 입력도 모두 같은 방식으로 처리
run_example("숫자가 아닌 여러 입력", ["3.14", "", "  ", "!!", "10", "1,000", "20"])

# 5) 음수처럼 int()로 변환이 되는 값은 에러 없이 그대로 계산
run_example("음수 입력", ["-10", "20"])


# 6) 직접 입력해서 실행해보고 싶다면 아래 주석을 해제하세요.
# add_numbers()
````

### 문제3. 카운트다운 타이머

- **문제 설명**  
  사용자로부터 시간(초)을 입력받아 해당 시간만큼 카운트 다운을 진행한 후, "타이머 종료!"를 출력하고 프로그램이 종료됩니다. 매 초마다 남은 시간을 화면에 출력합니다.

- **함수 설명**     
  
  `count_down(seconds: int)`:  
  - seconds: 카운트 다운할 시간(초)입니다.
  - 입력받은 초만큼 카운트 다운을 진행하며, 매 초마다 남은 시간을 출력합니다.
  - 시간이 종료되면 "타이머 종료!"를 출력하고 프로그램을 종료합니다.

- **입출력 예시**

  - 입력:
    ```
    타이머를 시작할 시간(초)을 입력하세요: 5
    ```

  - 출력:
    ```
    타이머 시작!
    5
    4
    3
    2
    1
    타이머 종료!
    ```

````python
# 문제
import time

def countdown_timer(seconds):
    print(f"타이머를 시작할 시간(초)을 입력하세요: {seconds}")

    print("타이머 시작!")
    while (seconds != 0):
        print(seconds)
        time.sleep(1)
        seconds -= 1
    print("타이머 종료!")
````

````python
# 프로그램 실행
countdown_timer(5)
````

### 문제1. 시간 추적 클래스(TimeTracker)

- **실습 설명**

  시간을 관리하고 추적하는 `TimeTracker` 클래스를 구현하는 프로젝트를 시작합니다. 시간 관리 기능은 특히 프로젝트 작업, 운동, 공부 시간 등 다양한 활동의 지속 시간을 측정하는 데 유용합니다.

  `TimeTracker` 클래스는 다음 기능을 제공해야 합니다:

  1. **시작 시간 설정**: 사용자가 활동을 시작할 때의 시간을 기록합니다.
  2. **종료 시간 설정**: 사용자가 활동을 종료할 때의 시간을 기록합니다.
  3. **경과 시간 계산**: 활동의 시작과 종료 사이의 시간 차이를 계산합니다.

  이 클래스의 인스턴스를 사용하여 각각의 활동에 대해 별도의 시간 추적을 할 수 있어야 합니다.

- **구현해야 할 메소드**

  - `start`: 현재 시간을 시작 시간으로 설정합니다.
  - `stop`: 현재 시간을 종료 시간으로 설정하고 경과 시간을 계산합니다.
  - `get_elapsed_time`: 마지막으로 기록된 시작 시간과 종료 시간 사이의 경과 시간을 분 단위로 반환합니다.

- **실습 결과 예시**

  다음과 같은 코드를 실행했을 때의 출력 예시입니다:

  ```python
  study_session = TimeTracker()
  study_session.start()
  # 1시간 30분 경과(가정)
  study_session.stop()

  print("활동 시간:", study_session.get_elapsed_time(), "분")
  ```

  예상 출력:

  ```
  공부한 시간: 90 분
  ```

- **요구 사항**

  1. 실제 시간을 추적하려면 Python의 `datetime` 모듈을 사용하여 현재 시간을 `datetime.now()`로 가져올 수 있습니다.
  2. 경과 시간은 분 단위로 반환해야 합니다.

````python
from datetime import datetime

# 문제
class TimeTracker:

    def start(self):
        self.start_time = datetime.now()

    def stop(self):
        self.stop_time = datetime.now()

    def get_elapsed_time(self):
        return (self.stop_time - self.start_time).seconds // 60
````

````python
# 사용 예시
# 활동 시작
study_session = TimeTracker()
study_session.start()
````

````python
# 활동 종료
study_session.stop()
print("공부한 시간:", study_session.get_elapsed_time(), "분")
````

- **실습 결과 예시**

    ```python
    voting_system = VoteSystem()
    ```

    ```
    Bob에게 투표하였습니다.
    투표 결과:
    Alice: 2 votes
    Bob: 1 vote
    Charlie: 0 votes
    ```

- **요구 사항**

  - 후보자는 중복 등록될 수 없습니다.
  - 등록되지 않은 후보자에게 투표할 수 없습니다.
  - 각 후보자의 이름과 투표 수는 사전(dictionary)을 사용하여 관리합니다.

````python
# 문제
class VoteSystem:
    candidate_votes = {}

    def add_candidate(self, name):
        if name in VoteSystem.candidate_votes: raise ValueError("이미 등록되어있는 후보자입니다.")
        VoteSystem.candidate_votes[name] = 0
        print(f"{name} 후보가 성공적으로 등록되었습니다.")

    def vote(self, name):
        if name not in VoteSystem.candidate_votes: raise ValueError("미등록 후보자입니다.") 
        VoteSystem.candidate_votes[name] += 1
        print(f"{name}에게 투표하였습니다.")

    @classmethod
    def get_results(cls):
        print("투표 결과:")
        for item in VoteSystem.candidate_votes.items():
            print(f"{item[0]}: {item[1]} votes")


````

- **구현해야 할 메소드**

  - `__init__`: 객체 생성 시 계좌 번호, 소유자 이름, 초기 잔액을 설정합니다.
  - `deposit`: 계좌에 금액을 입금합니다. 입금할 금액을 인자로 받고, 잔액을 업데이트합니다.
  - `withdraw`: 계좌에서 금액을 출금합니다. 출금할 금액을 인자로 받고, 잔액이 충분할 경우에만 출금을 허용하고 잔액을 업데이트합니다.

- **실습 결과 예시**

  - 예상 출력:

    ```
    김철수님의 계좌 123-456-789가 개설되었습니다. 초기 잔액: 100000원
    50000원이 입금되었습니다. 현재 잔액: 150000원
    20000원이 출금되었습니다. 현재 잔액: 130000원
    현재 잔액: 130000원
    ```


- **요구 사항**

  - 계좌에서 출금 시도 시 잔액보다 많은 금액을 출금하려고 하면, 출금을 거부하고 경고 메시지를 출력해야 합니다.
  - 모든 금액은 정수 또는 실수로 처리될 수 있어야 하며, 화폐 단위로만 입력받습니다.
  - 계좌 생성, 입금, 출금 및 잔액 조회 기능을 모두 구현해야 합니다.

````python
# 문제
class BankAccount:
    acc_balance = {}

    def __init__(self, acc_num, acc_hol, balance):
        if acc_num in BankAccount.acc_balance: raise ValueError("이미 개설한 계좌번호는 다시 개설할 수 없습니다.")
        self.__check_amount_validity(balance)

        self.my_acc_num = acc_num
        self.my_name = acc_hol
        # balance는 클래스 변수를 SSOT로 둔다. 구현에 따라 인스턴스 상태와 클래스 상태가 drift될 수도 있음.

        BankAccount.acc_balance[self.my_acc_num] = {"account_holder": self.my_name, "balance": balance}

        print(f"{self.my_name}님의 계좌 {acc_num}가 개설되었습니다. 초기 잔액: {BankAccount.acc_balance[self.my_acc_num]["balance"]}원")
    
    def deposit(self, amount):
        self.__check_amount_validity(amount)

        BankAccount.acc_balance[self.my_acc_num]["balance"] += amount

        print(f"{amount}원이 입금되었습니다. 현재 잔액: {BankAccount.acc_balance[self.my_acc_num]["balance"]}")

    def withdraw(self, amount):
        self.__check_amount_validity(amount)
        if BankAccount.acc_balance[self.my_acc_num]["balance"] < amount: raise ValueError("잔액보다 많은 양을 출금하려고 시도했습니다.")
        
        BankAccount.acc_balance[self.my_acc_num]["balance"] -= amount

        print(f"{amount}원이 출금되었습니다. 현재 잔액: {BankAccount.acc_balance[self.my_acc_num]["balance"]}")

    def get_balance(self):
        return BankAccount.acc_balance[self.my_acc_num]["balance"]

    def __check_amount_validity(self, amount):
        if not (isinstance(amount, int) or isinstance(amount, float)): raise TypeError("잔액은 정수 또는 실수로 입력해주세요.")
````

````python
# 사용 예시
my_account = BankAccount("123-456-789", "김철수", 100000)
my_account.deposit(50000)
my_account.withdraw(20000)
print(f"현재 잔액: {my_account.get_balance()}원")
````

````python
# 문제
from dateutil import parser # TODO: 날짜 포맷 정리하기

class ReservationSystem:
    reservations = {}

    def __init__(self, branch_name):
        # 이미 있는 지점을 다시 추가하려 할 때 예외 발생
        if branch_name in ReservationSystem.reservations: raise ValueError(f"{branch_name}지점은 이미 시스템 내에 존재합니다.")

        self.branch_name = branch_name

        # 예약 비어있는 상태로 신규 지점 생성
        ReservationSystem.reservations[self.branch_name] = []

    def add_reservation(self, customer_name, date, size):
        # 날짜 정규화
        parsed = parser.parse(date)

        ReservationSystem.reservations[self.branch_name].append({"customer_name": customer_name, "date": parsed.strftime("%Y-%m-%d"), "party_size": size})

    def list_reservations(self):

        print(f"{self.branch_name} 예약 목록:")

        for r in ReservationSystem.reservations[self.branch_name]:
            print(f"- {r["customer_name"]}, {r["date"]}, {r["party_size"]}명")

    @classmethod
    def sum_reservations(cls):
        return len(cls.reservations)
    
````

- **기능 설명**
  - 데이터 읽기: 사용자로부터 CSV 파일 경로를 입력받아 파일을 읽습니다. 파일은 각 할 일과 그에 소요되는 시간을 포함합니다.
  - 할 일 선택: 사용자에게 남은 시간을 입력받습니다. 이 시간을 기준으로 할 수 있는 최대한의 할 일을 선택합니다.
  - 결과 출력: 선택된 할 일을 소요 시간이 짧은 순으로 정렬하여 출력합니다. 출력 형식은 목록 형태로 할 일과 예상 소요 시간을 포함합니다.

- **입출력 예시**

    - 할 일 및 소요 시간 CSV 파일 예시 (`tasks.csv`):
        ```
        할일,소요시간
        공부하기,120
        운동하기,60
        가계부정리,30
        영화보기,150
        ```

    - 입력:
        ```
        남은 시간(분): 210
        ```

    - 출력:
        ```
        선택된 할 일 목록:
        1. 가계부정리 - 예상 소요 시간: 30분
        2. 운동하기 - 예상 소요 시간: 60분
        3. 공부하기 - 예상 소요 시간: 120분
        ```

````python
### 문제를 위한 파일 생성코드입니다. 문제를 풀기 전 이 코드를 반드시 실행해주세요.

import csv
# (시뮬레이션을 위한) 예시 파일 작성코드
def write_tasks_to_csv(filename, tasks):
    with open(filename, 'w', newline='') as csvfile:
        fieldnames = ['할일', '소요시간']
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames)

        writer.writeheader()
        for task, time in tasks:
            writer.writerow({'할일': task, '소요시간': time})

# 할 일과 소요 시간 데이터
tasks = [
    ('공부하기', 120),
    ('운동하기', 60),
    ('가계부정리', 30),
    ('영화보기', 150)
]

# CSV 파일로 저장
write_tasks_to_csv('tasks.csv', tasks)
````

````python
import csv

# 문제
def load_tasks(filename):
    tasks_and_time = []

    with open(filename, "r", newline='', encoding='utf-8') as csvfile:
        for row in csv.DictReader(csvfile):
            tasks_and_time.append((row['할일'], int(row['소요시간'])))

    return tasks_and_time

# 가장 많은 일을 할 수 있는 조합
def suggest_tasks(tasks, remaining_time):

    print("선택된 할 일 목록: ")
    tasks_sorted_by_time_asc = sorted(tasks, key=lambda t: t[1])

    print(f"정렬된 리스트: {tasks_sorted_by_time_asc}")

    cnt = 1
    remaining = remaining_time
    for task in tasks_sorted_by_time_asc:
        
        if task[1] > remaining:
            return
    
        print(f"{cnt}. {task[0]} - 예상 소요 시간: {task[1]}분")

        remaining -= task[1]
        cnt += 1

def main():
    try:    
        remaining_time = int(input("남은 시간(분): "))

        tasks = load_tasks('tasks.csv')
        suggest_tasks(tasks, remaining_time)

    except ValueError:
        raise ValueError("남은 시간(분)을 숫자로만 입력해주세요. ex: 120")
````

### 문제2. 수학 퍼즐: 소수의 합

- **실습 설명**  
  이 프로그램은 사용자로부터 어떤 수 이하의 소수의 합을 구할 것인지 입력 받아, 해당 수까지의 모든 소수의 합을 계산하여 출력합니다. 소수(prime number)는 1과 자기 자신만을 약수로 가지는 자연수를 말합니다. 소수를 찾는 방법은 자유롭게 선택할 수 있지만, 효율적인 방법을 고려하여 구현해야 합니다.

- **기능 설명**
  - 소수 판별: 주어진 수까지의 모든 자연수에 대해 소수 여부를 판별합니다. (힌트 : 에라토스테네스의 체)
  - 소수 합 계산: 판별된 소수들의 합을 계산합니다.
  - 결과 출력: 계산된 소수의 합을 출력합니다. 출력 형식은 사용자가 입력한 수까지의 소수 합과 그 값을 포함합니다.

- **입출력 예시**

  - 입력:

  ```
  어떤 수까지의 소수의 합을 구하시겠습니까?: 10
  ```
  - 출력:

  ```
  10 이하 소수의 합은 17입니다.
  ```

````python
# 문제
def get_list_of_primes_under(n):
    primes = [True] * (n + 1)
    primes[0] = primes[1] = False
    
    # n의 제곱근까지만 확인
    for i in range(2, int(n**0.5) + 1):
        if primes[i]:
            # i의 배수들을 False로 변환 (i * i부터 시작)
            for j in range(i * i, n + 1, i):
                primes[j] = False
                
    # True로 남아있는 인덱스(소수)들만 리스트로 반환
    return [i for i in range(n + 1) if primes[i]]

def sum_of_primes_under(max):
    return sum(get_list_of_primes_under(max))

def main():
    try:
        max = int(input("어떤 수까지의 소수의 합을 구하시겠습니까?: "))

        print(f"{max} 이하 소수의 합은 {sum_of_primes_under(max)}입니다.")
    except (ValueError):
        raise ValueError("숫자만 입력해주세요. ex: 10")
````

````python
# 문제
from datetime import datetime, timedelta
import uuid

# -------------------------------- #
# ------ Custom Error Class ------ #
# -------------------------------- #

class BookNotAvailableError(Exception):
    """요청한 도서가 도서 목록에 없거나 대여 가능한 수량이 없을 때 발생하는 예외."""

    def __init__(self, isbn):
        """문제가 된 isbn을 담은 예외 메시지를 만든다.

        Args:
            isbn: 도서 목록에 없거나 수량이 부족한 도서의 isbn.
        """
        super().__init__(f"해당 isbn({isbn}) 도서가 도서 목록에 없거나 수량이 부족합니다.")

class MemberNotExistsError(Exception):
    """도서관에 등록되지 않은 회원이 도서를 대여하려 할 때 발생하는 예외."""

    def __init__(self, name, uuid):
        """등록되지 않은 회원의 정보를 담은 예외 메시지를 만든다.

        Args:
            name: 회원 이름.
            uuid: 회원 번호(uuid).
        """
        super().__init__(f"회원({name}, {uuid})이 회원 목록에 없습니다.")

class MemberHaveNotRentThisBookError(Exception):
    """회원이 대여한 적 없는 도서를 반납하려 할 때 발생하는 예외."""

    def __init__(self, member_name, isbn):
        """반납을 시도한 회원과 도서 정보를 담은 예외 메시지를 만든다.

        Args:
            member_name: 반납을 시도한 회원의 이름.
            isbn: 반납을 시도한 도서의 isbn.
        """
        super().__init__(f"멤버({member_name})은 해당 도서 (isbn: {isbn})을 대여한 적이 없습니다.")

class RentalNotInSystemError(Exception):
    """회원의 대여 기록은 있으나 시스템 대여 목록에는 없을 때 발생하는 예외."""

    def __init__(self, rental_uuid):
        """시스템 대여 목록에서 찾지 못한 대여 번호를 담은 예외 메시지를 만든다.

        Args:
            rental_uuid: 시스템 대여 목록에 없는 대여 기록의 uuid.
        """
        super().__init__(f"대여({rental_uuid})는 시스템 대여 목록에 없습니다.")


# ------------------------------- #
# ---------- Book Class --------- #
# ------------------------------- #
class Book:
    """도서를 나타내는 클래스.
    """

    def __init__(self, title: str, author: str, publication_year: int, isbn: str):
        """생성자 메서드.

        전달받은 도서 정보를 저장하고 수량(quantity)을 1로 초기화한다.

        Args:
            title: 책의 제목.
            author: 책의 저자.
            publication_year: 출간 연도.
            isbn: isbn. 도서의 식별자로 사용.
        """
        self.title = title
        self.author = author
        self.publication_year = publication_year
        self.isbn = isbn
        self.quantity = 1

    def __str__(self):
        """도서 정보를 한 줄 문자열로 만든다.

        Returns:
            str: "제목 (저자: OOO, 출판 년도: OOOO, 수량: N)" 형식의 문자열.
        """
        return f"{self.title} (저자: {self.author}, 출판 년도: {self.publication_year}, 수량: {self.quantity})"

# ------------------------------- #
# --------- Member Class -------- #
# ------------------------------- #
class Member:
    """ 회원을 나타내는 클래스.
    """

    def __init__(self, name: str):
        """생성자 메서드.

        이름을 인자로 전달하면 회원 번호(uuid)를 생성하고 대여 목록을 빈 리스트로 초기화한다.

        Args:
            name: 회원 이름.
        """
        self.name = name
        self.uuid = uuid.uuid4()
        self.rental_list = [] # [Rental, Rental, ...]

    def rent_a_book(self, rental):
        """회원의 대여 목록에 대여 기록을 추가한다.

        Args:
            rental: 이 회원이 도서를 대여하며 생성된 Rental 객체.

        Returns:
            None
        """
        self.rental_list.append(rental)

    def __str__(self):
        """회원 정보를 한 줄 문자열로 만든다.

        Returns:
            str: "이름: OOO, uuid: OOO" 형식의 문자열.
        """
        return f"이름: {self.name}, uuid: {self.uuid}"

# ------------------------------- #
# --------- Rental Class -------- #
# ------------------------------- #
class Rental:
    """ 도서 한 건의 대여 기록을 나타내는 클래스.
    """

    def __init__(self, book: Book, member: Member):
        """생성자 메서드.

        도서 객체와 회원 객체를 전달하면 대여 번호(uuid)를 생성하고,
        대여일을 오늘 날짜로, 반납 예정일을 대여일로부터 14일 뒤로 설정한다.

        Args:
            book: 대여하는 Book 객체.
            member: 대여하는 Member 객체.
        """
        self.uuid = uuid.uuid4()
        self.book = book
        self.member = member
        self.rental_date = self.get_now()
        self.return_date = self.rental_date + timedelta(days=14)

    @staticmethod
    def get_now():
        """오늘 날짜를 구한다.

        Returns:
            datetime.date: 메서드를 호출한 시점의 날짜.
        """
        return datetime.now().date()

# -------------------------------- #
# --- Library Management Class --- #
# -------------------------------- #
class LibraryManagement:
    """ 도서관 관리 클래스. 소장 도서 리스트, 회원 리스트, 대여 기록을 관리한다.
    """
    def __init__(self):
        """생성자 메서드.

        도서 목록, 회원 목록, 대여 목록을 각각 빈 딕셔너리로 초기화한다.

        Attributes:
            book_dict: {isbn: Book} 형태의 소장 도서 목록.
            member_dict: {uuid: Member} 형태의 등록 회원 목록.
            rental_dict: {uuid: Rental} 형태의 시스템 대여 목록.
        """
        self.book_dict = dict()   # {isbn: Book, isbn: Book ...}
        self.member_dict = dict()   # {uuid: Member, uuid: Member ...}
        self.rental_dict = dict()       # {uuid: Rental, uuid: Rental ...}

    # 도서 추가
    def add_book(self, title: str, author: str, publication_year: int, isbn: str) -> None:
        """도서관에 도서 한 권을 추가한다.

        책이 목록에 없으면 딕셔너리에 요소를 추가하고, 이미 있으면 수량을 하나 늘린다.

        Args:
            title: 책의 제목.
            author: 책의 저자.
            publication_year: 출간 연도.
            isbn: isbn. 도서의 식별자로 사용.

        Returns:
            None
        """
        if self.__book_exists(isbn):
            # 책이 목록에 있으면 수량만 늘린다.
            book = self.book_dict[isbn]
            book.quantity += 1
        else:
            # 책이 목록에 없으면 새로 추가한다.
            new_book = Book(title, author, publication_year, isbn)
            self.book_dict[isbn] = new_book

        print(f"'{title}' (저자: {author}, 출판년도: {publication_year}) 도서가 추가되었습니다.")

    # 도서 삭제
    def del_book(self, isbn: str):
        """도서관에서 도서 한 권을 삭제한다.

        수량이 2권 이상인 도서는 수량만 줄어들고, 수량이 1권인 도서는 도서 목록에서 아예 제거된다.

        Args:
            isbn: 도서 isbn.

        Returns:
            Book: 삭제된 Book 객체. 수량이 남아있는 경우 남아있는 수량이 보이고, 없는 경우 0으로 보인다.

        Raises:
            BookNotAvailableError: 해당 isbn 도서가 목록에 없거나 수량이 0일 때.
        """

        self.__check_if_the_book_available(isbn)

        deleted_book = self.book_dict[isbn] # Book 객체

        deleted_book.quantity -= 1 # 수량 1개 줄이기

        if deleted_book.quantity <= 0: del self.book_dict[isbn] # 수량이 0개 이하면 딕셔너리에서 삭제

        return deleted_book

    # 도서 목록 조회
    def print_books(self):
        """도서관이 소장중인 모든 도서 목록을 출력한다.

        소장중인 도서가 하나도 없으면 안내 문구를 대신 출력한다.

        Returns:
            None
        """

        if not self.book_dict:
            print("소장중인 도서가 없습니다.")
            return

        print("도서 목록:")
        for book in self.book_dict.values():
            print(f"- {book}")

    # 회원 등록
    def add_member(self, name: str) -> Member:
        """도서관에 새로운 회원을 등록한다.

        회원은 이름이 아니라 내부적으로 생성되는 uuid로 구분하므로 동명이인도 등록될 수 있다.

        Args:
            name: 새로운 회원의 이름.

        Returns:
            Member: 회원 번호(uuid)가 부여되어 회원 목록에 등록된 Member 객체.
        """

        new_member = Member(name)
        self.member_dict[new_member.uuid] = new_member

        print(f"회원 '{name}'님이 등록되었습니다.")

        return new_member

    # 회원 목록 조회
    def print_members(self):
        """도서관에 등록된 회원 목록을 출력한다.

        등록된 회원이 하나도 없으면 안내 문구를 대신 출력한다.

        Returns:
            None
        """

        if not self.member_dict:
            print("도서관에 등록된 회원이 없습니다.")
            return

        print("회원 목록:")
        for member in self.member_dict.values():
            print(f"- {member}")

    # 도서 대여
    def rent_book(self, isbn: str, member: Member) -> None:
        """회원에게 도서 한 권을 대여한다.

        도서 수량을 하나 줄이고, 생성한 대여 기록을 회원의 대여 목록과 시스템 대여 목록 양쪽에 추가한다.
        수량이 1권이었던 도서는 목록에서 삭제하지 않고 수량만 0으로 만든다.

        Args:
            isbn: 대여할 도서의 isbn.
            member: 대여하는 Member 객체.

        Returns:
            None

        Raises:
            BookNotAvailableError: 해당 isbn 도서가 목록에 없거나 수량이 0일 때.
            MemberNotExistsError: 도서관에 등록되지 않은 회원일 때.
        """
        # 빌리려는 책이 목록에 없거나 수량이 0이면 에러
        self.__check_if_the_book_available(isbn)
        # 회원이 등록되어있지 않으면 에러
        self.__check_if_the_member_exists(member)

        # 수량 --. 수량이 1개였다면 삭제하지 않고 수량을 0으로 만든다.
        book = self.book_dict[isbn]
        book.quantity -= 1

        # 대여 객체 생성
        rental = Rental(book, member)

        # 멤버 객체의 대여 목록에 추가.
        self.member_dict[member.uuid].rent_a_book(rental)

        # 시스템 대여 목록에 추가.
        self.rental_dict[rental.uuid] = rental

        print(f"'{member.name}' 회원님이 '{book.title}'을 대여하였습니다.")


    # 도서 반납
    def return_book(self, isbn: str, member: Member) -> None:
        """회원이 대여한 도서를 반납한다.

        대여 기록을 회원과 시스템 양쪽에서 제거하고 도서 수량을 하나 복구한다.

        Args:
            isbn: 반납할 도서의 isbn.
            member: 반납하는 Member 객체.

        Returns:
            None

        Raises:
            MemberHaveNotRentThisBookError: 회원의 대여 목록에 해당 도서의 대여 기록이 없을 때.
            RentalNotInSystemError: 시스템 대여 목록에 해당 대여 기록이 없을 때.
        """

        # 멤버 객체의 대여 리스트에 그 isbn 책이 있는 지 검사.
        rental_obj = None
        for rental in member.rental_list:
            if rental.book.isbn == isbn and rental.member is member:
                rental_obj = rental
                break

        # 없으면 에러
        if rental_obj is None: raise MemberHaveNotRentThisBookError(member.name, isbn)

        # 도서관 시스템 렌탈 리스트에 그 uuid가 있는 지 검사. 없으면 에러.
        if rental_obj.uuid not in self.rental_dict: raise RentalNotInSystemError(rental_obj.uuid)

        # 검사를 모두 통과한 뒤에 상태를 변경한다. (중간에 에러가 나도 대여 기록이 훼손되지 않도록)
        member.rental_list.remove(rental_obj)   # 회원의 대여 목록에서 제거
        del self.rental_dict[rental_obj.uuid]   # 시스템 대여 목록에서 제거
        self.book_dict[isbn].quantity += 1      # 도서 개수 하나 늘리기

        print(f"'{member.name}' 회원님이 '{rental_obj.book.title}' 도서를 반납하였습니다.")

    # ------------------------------- #
    # ---- private helper method ---- #
    # ------------------------------- #

    # 해당 name의 멤버가 없으면 에러를 뿜는다
    def __check_if_the_member_exists(self, member: Member):
        """회원이 도서관에 등록되어 있는지 검사한다.

        Args:
            member: 검사할 Member 객체.

        Returns:
            None

        Raises:
            MemberNotExistsError: 해당 회원이 회원 목록에 없을 때.
        """
        if not self.__member_exists(member.uuid):
            raise MemberNotExistsError(member.name, member.uuid)

    # 해당 isbn 도서가 없거나 수량이 0이면 에러를 뿜는다.
    def __check_if_the_book_available(self, isbn: str):
        """도서가 대여 가능한 상태인지 검사한다.

        Args:
            isbn: 검사할 도서의 isbn.

        Returns:
            None

        Raises:
            BookNotAvailableError: 해당 isbn 도서가 목록에 없거나 수량이 0일 때.
        """
        if not self.__book_exists(isbn):
            raise BookNotAvailableError(isbn)


    def __book_exists(self, isbn: str) -> bool:
        """도서 목록에 해당 isbn 도서가 남아 있는지 확인한다.

        Args:
            isbn: 도서 isbn.

        Returns:
            bool: self.book_dict에 isbn 키가 존재하고 수량이 0보다 크면 True,
                키가 존재하지 않거나 수량이 0 이하면 False.
        """
        return (isbn in self.book_dict) and (self.book_dict[isbn].quantity > 0)

    def __member_exists(self, uuid: uuid) -> bool:
        """회원 목록에 해당 uuid 회원이 있는지 확인한다.

        Args:
            uuid: 회원 uuid.

        Returns:
            bool: self.member_dict에 uuid가 존재하면 True, 존재하지 않으면 False.
        """
        return uuid in self.member_dict
````

````python
# 사용 예시
library = LibraryManagement()
print("----------------------------------------")
library.add_book("1984", "조지 오웰", 1949, "978-0451524935")
print("----------------------------------------")
library.add_book("앵무새 죽이기", "하퍼 리", 1960, "978-0446310789")
print("----------------------------------------")
library.print_books()
print("----------------------------------------")
gildong = library.add_member("홍길동")

print("----------------------------------------")
library.rent_book("978-0451524935", gildong)
print("----------------------------------------")
library.print_books()
print("----------------------------------------")
library.print_members()
print("----------------------------------------")
library.return_book("978-0451524935", gildong)
print("----------------------------------------")
library.print_books()
print("----------------------------------------")
library.print_members()
````

심화문제를 풀고 싶으나 어디서부터 시작해야 할지 막막하다면, 이래의 단계별 개발 가이드를 참고해보세요.


- **1단계: 기본 클래스 정의**
  - **목표**: 도서, 회원, 대여 정보를 저장할 기본 클래스를 생성합니다.

  - **주요 작업**
    1. `Book` 클래스 생성: `__init__` 메소드에는 `title`, `author`, `publication_year`, `isbn` 파라미터를 포함시킵니다.
    2. `Member` 클래스 생성: `__init__` 메소드에는 `name` 파라미터를 포함시키고, 대여 중인 도서 목록을 관리할 리스트를 초기화합니다.
    3. `Rental` 클래스 생성: `__init__` 메소드에는 `book`, `member`, `rental_date`, `return_date` 파라미터를 포함시킵니다.

- **2단계: 관리 시스템 클래스 구현**
  - **목표**: 도서, 회원, 대여 정보를 관리하는 메소드를 포함하는 `LibraryManagement` 클래스를 구현합니다.

  - **주요 작업**
    1. `LibraryManagement` 클래스에 필요한 인스턴스 변수 초기화: 도서 목록, 회원 목록, 대여 목록.
    2. 도서 추가 메소드(`add_book`) 구현.
    3. 회원 추가 메소드(`add_member`) 구현.
    4. 도서 대여 메소드(`rent_book`) 구현.
    5. 도서 반납 메소드(`return_book`) 구현.

- **3단계: 도서 및 회원 정보 조회 기능 추가**
  - **목표**: 도서 및 회원 정보를 조회하고 출력하는 기능을 구현합니다.

  - **주요 작업**
    1. 도서 목록 출력 메소드(`print_books`) 구현.
    2. 회원 목록 출력 메소드(`print_members`) 구현.

- **4단계: 테스트 및 디버깅**
  - **목표**: 관리 시스템의 테스트하여 오류를 찾고 수정합니다.

  - **주요 작업**
    1. 각 단계에서 구현한 기능을 개별적으로 테스트합니다.
    2. 예상하지 못한 에러가 발생하면 원인을 분석하고 수정합니다.
    3. 오류 메시지 외에 처리 과정을 확인하고 필요에 따라 수정합니다.


````python
# 예외 상황 테스트
# 구현한 예외 처리가 의도대로 동작하는지 하나씩 확인한다.

def check_error(expected_error, description, func, *args, **kwargs):
    """ func 실행 시 expected_error가 발생하는지 검사하고 결과를 출력한다.

    Args:
        expected_error: 기대하는 예외 클래스
        description: 테스트 설명
        func: 실행할 함수(메서드)
        *args, **kwargs: func에 전달할 인자

    Returns:
        None
    """
    print(f"[테스트] {description}")
    try:
        func(*args, **kwargs)
    except expected_error as e:
        print(f"  -> 통과: {type(e).__name__} 발생 | {e}")
    except Exception as e:
        print(f"  -> 실패: {expected_error.__name__}를 기대했지만 {type(e).__name__} 발생 | {e}")
    else:
        print(f"  -> 실패: {expected_error.__name__}가 발생하지 않음")
    print("----------------------------------------")


test_lib = LibraryManagement()

# =============================================== #
# 1. 도서/회원이 하나도 없을 때의 목록 조회
# =============================================== #
print("### 1. 빈 목록 조회 ###")
test_lib.print_books()      # "소장중인 도서가 없습니다."
test_lib.print_members()    # "도서관에 등록된 회원이 없습니다."
print("----------------------------------------")

# =============================================== #
# 2. BookNotAvailableError
#    : 도서 목록에 없거나 수량이 0인 도서에 접근할 때
# =============================================== #
print("### 2. BookNotAvailableError ###")

# 2-1. 목록에 없는 isbn을 대여
test_member = test_lib.add_member("홍길동")
check_error(BookNotAvailableError, "등록되지 않은 isbn 도서를 대여",
            test_lib.rent_book, "없는-isbn", test_member)

# 2-2. 목록에 없는 isbn을 삭제
check_error(BookNotAvailableError, "등록되지 않은 isbn 도서를 삭제",
            test_lib.del_book, "없는-isbn")

# 2-3. 수량이 0이 된 도서를 대여 (재고 소진)
test_lib.add_book("1984", "조지 오웰", 1949, "978-0451524935")   # 수량 1
test_lib.rent_book("978-0451524935", test_member)               # 수량 0
test_member2 = test_lib.add_member("김철수")
check_error(BookNotAvailableError, "재고가 0인 도서를 다른 회원이 대여",
            test_lib.rent_book, "978-0451524935", test_member2)

# =============================================== #
# 3. MemberNotExistsError
#    : 도서관에 등록되지 않은 회원이 도서를 대여할 때
# =============================================== #
print("### 3. MemberNotExistsError ###")

test_lib.add_book("앵무새 죽이기", "하퍼 리", 1960, "978-0446310789")
stranger = Member("이방인")   # add_member를 거치지 않아 member_dict에 없는 회원
check_error(MemberNotExistsError, "등록되지 않은 회원이 도서를 대여",
            test_lib.rent_book, "978-0446310789", stranger)

# =============================================== #
# 4. MemberHaveNotRentThisBookError
#    : 대여한 적 없는 도서를 반납할 때
# =============================================== #
print("### 4. MemberHaveNotRentThisBookError ###")

# 4-1. 대여 목록이 비어 있는 회원의 반납
check_error(MemberHaveNotRentThisBookError, "대여 기록이 전혀 없는 회원이 반납",
            test_lib.return_book, "978-0446310789", test_member2)

# 4-2. 다른 도서만 대여한 회원이 대여하지 않은 도서를 반납
test_member3 = test_lib.add_member("이영희")
test_lib.rent_book("978-0446310789", test_member3)   # '앵무새 죽이기'만 대여
check_error(MemberHaveNotRentThisBookError, "다른 도서만 대여한 회원이 대여하지 않은 도서를 반납",
            test_lib.return_book, "978-0451524935", test_member3)

# 4-3. 2권을 대여한 회원이 그중 1권만 반납 (예외 없이 반납되어야 하는 경우)
test_member5 = test_lib.add_member("정우성")
test_lib.add_book("1984", "조지 오웰", 1949, "978-0451524935")        # 재고 보충
test_lib.add_book("앵무새 죽이기", "하퍼 리", 1960, "978-0446310789")   # 재고 보충
test_lib.rent_book("978-0451524935", test_member5)
test_lib.rent_book("978-0446310789", test_member5)
print("[테스트] 2권을 대여한 회원이 그중 1권만 반납")
try:
    test_lib.return_book("978-0451524935", test_member5)
except Exception as e:
    print(f"  -> 실패: 예외 없이 반납되어야 하지만 {type(e).__name__} 발생 | {e}")
else:
    print(f"  -> 통과: 반납 완료, 남은 대여 {len(test_member5.rental_list)}건")
print("----------------------------------------")

# =============================================== #
# 5. RentalNotInSystemError
#    : 회원의 대여 기록은 있지만 시스템 대여 목록에는 없을 때
# =============================================== #
print("### 5. RentalNotInSystemError ###")

test_member4 = test_lib.add_member("박지성")
test_lib.add_book("데미안", "헤르만 헤세", 1919, "978-8937462498")
test_lib.rent_book("978-8937462498", test_member4)
test_lib.rental_dict.clear()   # 시스템 대여 목록만 인위적으로 비워 상황을 재현
check_error(RentalNotInSystemError, "시스템 대여 목록에 없는 대여를 반납",
            test_lib.return_book, "978-8937462498", test_member4)

# =============================================== #
# 6. 동명이인 등록 정책
#    : 회원은 이름이 아니라 uuid로 구분하므로 동명이인 등록을 허용한다.
# =============================================== #
print("### 6. 동명이인 등록 ###")

same_name_member = test_lib.add_member("홍길동")   # 이미 '홍길동'이 등록되어 있음
print(f"  -> 통과: 이름이 같아도 서로 다른 회원으로 등록됨 "
      f"({test_member.uuid} != {same_name_member.uuid})")
print("----------------------------------------")

# =============================================== #
# 7. 예외는 아니지만 별도로 처리한 경계 상황
# =============================================== #
print("### 7. 경계 상황 처리 ###")

edge_lib = LibraryManagement()

# 7-1. 같은 isbn을 중복 추가하면 새 항목이 아니라 수량이 증가
edge_lib.add_book("1984", "조지 오웰", 1949, "978-0451524935")
edge_lib.add_book("1984", "조지 오웰", 1949, "978-0451524935")
edge_lib.print_books()   # 수량: 2
print("----------------------------------------")

# 7-2. 수량이 2권 이상이면 삭제 시 수량만 감소
deleted = edge_lib.del_book("978-0451524935")
print(f"삭제 결과: {deleted}")   # 수량: 1
edge_lib.print_books()
print("----------------------------------------")

# 7-3. 마지막 1권을 삭제하면 도서 목록에서 완전히 제거
deleted = edge_lib.del_book("978-0451524935")
print(f"삭제 결과: {deleted}")   # 수량: 0
edge_lib.print_books()           # "소장중인 도서가 없습니다."
print("----------------------------------------")

# 7-4. 대여로 수량이 0이 된 도서는 목록에는 남되 대여는 불가
edge_lib.add_book("데미안", "헤르만 헤세", 1919, "978-8937462498")
edge_member = edge_lib.add_member("최수현")
edge_lib.rent_book("978-8937462498", edge_member)
edge_lib.print_books()   # 수량: 0 으로 목록에 남아 있음
check_error(BookNotAvailableError, "수량 0인 도서를 삭제",
            edge_lib.del_book, "978-8937462498")

# 7-5. 반납하면 수량이 복구되어 다시 대여 가능
edge_lib.return_book("978-8937462498", edge_member)
edge_lib.print_books()   # 수량: 1
````
