```python
p15_labels = torch.tensor([0, 1, 2])
p15_scores = torch.tensor([0.1, 0.8, 0.3])
p15_total = p15_labels * p15_scores

print("before:", p15_labels.dtype, p15_scores.dtype, "→", p15_total.dtype)

p15_labels_float = None  # 관찰을 위해 새 Tensor를 만들어봄

if p15_labels_float is not None:
    p15_total_float = p15_labels_float * p15_scores
    print("after: ", p15_labels_float.dtype, p15_scores.dtype, "→", p15_total_float.dtype)
else:
    print("labels.float() 결과를 별도 변수에 담아 비교함")
```
