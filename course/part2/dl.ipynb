### 짧은 실습 · 습도만 선택하기

**목적**: axis 의미를 이용해 slicing함.

`weather`에서 모든 관측의 습도만 선택하고 shape를 확인함.


```python
weather = torch.tensor([
    [18.0, 45.0, 0.0],
    [21.0, 55.0, 1.0],
    [25.0, 60.0, 1.0],
    [20.0, 50.0, 0.0],
])
feature_names = ["온도", "습도", "비 여부"]
print(weather)
print("shape:", tuple(weather.shape))
```

```python
only_humidity = weather[:, 1]
print(only_humidity)
print('shape: ', tuple(only_humidity.shape))
```

### 짧은 실습 · shape guard 통과시키기

아래 `target_attempt`을 수정해 prediction과 같은 shape로 만듦. 실행 결과가 `안전한 1:1 비교 준비됨`이 되면 완료함.

```python
prediction_column = torch.tensor([[2.0], [4.0], [6.0], [8.0]]) # [4,1]
target_flat = torch.tensor([3.0, 5.0, 7.0, 9.0]) # [4]

print(f"prediction_column shape: {prediction_column.shape}")
print(f"target_flat shape: {target_flat.shape}")

# 1번 인덱스에 차원을 추가: [4] -> [4,1] 
target_attempt = target_flat.unsqueeze(1) # [4,1]

print(f"target_attempt shape: {target_attempt.shape}")

if target_attempt.shape == prediction_column.shape:
    print("안전한 1:1 비교 준비됨")
else:
    print("shape를 다시 확인함:", tuple(prediction_column.shape), tuple(target_attempt.shape))
```
