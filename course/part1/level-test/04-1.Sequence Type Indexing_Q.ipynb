## 4. Sequence Type Indexing💻
***
#### **`연습예제`** 💻

`아래 리스트(List)에서` **`index 함수`** `를 포함한 다양한 방법으로` **`"Banana"`** `를 인덱싱(추출) 해보세요.` `출력 결과와 같아야 되요.`


```python
# 직접 접근, index함수 사용 가능
x = ['Orange', 'Cherry', 'Apple', 'Kiwi', 'Banana', 'Strawberry']

출력 결과 : Banana
```

#### **`코딩영역`**

````python
# 코딩 영역

x = ['Orange', 'Cherry', 'Apple', 'Kiwi', 'Banana', 'Strawberry']


# 1. index 직접 접근
print(x[4])

# 2. index()로 index 추출 후 접근
idx = x.index('Banana')
print(x[idx])

# 3. index() 인자로 리스트 안에 없는 값을 넣으면 어떨게 될까?
# x.index('Grape') # ValueError: 'Grape' is not in list 

# in 연산자를 사용하면 리스트 안에 특정 값이 있는 지 bool로 확인할 수 있습니다. 
print('Grape' in x) # False

# in으로 안전장치를 걸어봅시다
if 'Banana' in x:
    idx = x.index('Banana')
    print(x[idx])


# TODO:
# 1. index()의 구현
# 2. in의 구현
````
