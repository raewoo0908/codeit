#### **`연습예제1`** 💻

`아래 파이썬 코드가 왜` **`에러(예외)`**  `가 발생하는지 생각해보고` **`동작하도록 수정`** `후 코딩영역에 작성하세요.`

```python
x = "Seoul"
y = 25
z = x + y

print(f'x + y : {z}')
```

#### **`코딩영역1`**

````python
# 코딩 영역

# 발생 에러: can only concatenate str (not "int") to str
# 에러 발생 이유: str과 int는 서로 다른 자료형이기 때문에 연산이 불가능합니다. 따라서 str을 int로 변환하거나 int를 str로 변환해야 합니다.
# 더 구체적으로: + 연산자의 구현은?

x = "Seoul"
y = 25
z = x + str(y)

print(f'x + y : {z}')
````
