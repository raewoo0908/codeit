## 13. Dictionary Data Filtering💻
***
#### **`연습예제`** 💻

`아래와 같은` **`딕셔너리`** `구조에서` **`Value 값이 25 이상`** `필터링 후 출력하세요.` **`다양한 방법`** `으로 코딩하세요.`



```python
d = {'a': 8, 'b': 33, 'c': 15, 'd': 26, 'e': 12, 'f': 120}
출력 결과 : {'b': 33, 'd': 26, 'f': 120}
```

#### **`코딩영역`**

````python
# 코딩 영역

d = {'a': 8, 'b': 33, 'c': 15, 'd': 26, 'e': 12, 'f': 120}


print(dict(filter(lambda x: x[1] >= 25, d.items())))

# items()의 결과는 dict_items([('a', 8), ('b', 33), ('c', 15), ('d', 26), ('e', 12), ('f', 120)]) 이고, 각각의 요소는 튜플로 구성되어있다. 
# 따라서 lambda에서 x[1]을 통해 튜플의 두 번째 요소를 가져와 25 이상인 경우만 필터링한다.
````
