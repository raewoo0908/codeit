## 5. Sequence Type Slicing💻
***
#### **`연습예제`** 💻

`아래 리스트(List)에서` **`슬라이싱을 사용`** `해서` **`[e,f,g]`** `값을 출력해보세요. 다양한 슬라이싱 방법으로 시도해보세요.`



```python
x = ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j", "k", "l", "m"]

출력 결과 : [e,f,g]
```

#### **`코딩영역`**

````python
# 코딩 영역

x = ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j", "k", "l", "m"]

# index 직접 접근
print(x[4:7])

# index()로 index 추출 후 접근
idx1 = x.index("e")
idx2 = x.index("g")
print(x[idx1:idx2 + 1])

# 음수 인덱스로 접근하기
print(x[-9:-6])
print(x[-9:7])
print(x[4:-6])

# slicing 세번째 인자 넣기
print(x[4:10:2]) # start:end:step

# TODO: 슬라이싱의 구현
````
