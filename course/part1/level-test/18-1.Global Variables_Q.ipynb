## 18. Global Variables💻
***
#### **`연습예제`** 💻

`아래` **`함수가 실행 시 에러가 발생하는 이유`** `를 생각해보세요.` **`함수 수정 후`** `정상적으로 실행해보세요.`  **`결과값 : 1000`**



```python
# 전역변수
x = 100

def test():
    x = x * 10
    return x

print(test())
```

#### **`코딩영역`**

````python
# 코딩 영역
# error: UnboundLocalError: cannot access local variable 'x' where it is not associated with a value
# 발생 이유: 변수 x에 접근하려고 하는데, 함수 안에서 정의되어있지 않기 때문에 발생하는 에러입니다. 함수 안에서 x를 사용하려면, 함수 안에서 x를 정의하거나, global 키워드를 사용하여 전역 변수를 참조해야 합니다.
# 네임스페이스와 연관된 설명: TODO

x = 100

def test():
    global x 
    x = x * 10
    return x

print(test())
````
