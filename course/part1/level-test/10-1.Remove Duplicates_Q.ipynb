## 10. Remove Duplicates💻
***
#### **`연습예제`** 💻

`아래와 리스트에서` **`중복`** `되는` **`원소`** `를` `제거 후 출력하세요.`  **`다양한 방법`** `코딩하세요.` 



```python
x = ["a", 1, "b", 2, "a", 3, "b", 4, 5, "b"]
출력 결과 : [1, 2, 3, 4, 5, 'a', 'b']
```

#### **`코딩영역`**

````python
# 코딩 영역

x = ["a", 1, "b", 2, "a", 3, "b", 4, 5, "b"]


num_set = set(filter(lambda i: isinstance(i, int), x))

str_set = set(filter(lambda i: isinstance(i, str), x))

list(num_set.union(str_set))

````
