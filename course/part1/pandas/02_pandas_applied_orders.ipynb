# Analysis-Ready Gate

이제 raw 1,000행에서 분석에 사용할 논리 주문으로 상태를 전환합니다.

`orders`는 원본에서 시작한 작업 데이터이고, `logical_orders`는 중복과 충돌 판단을 반영한 분석용 데이터입니다.

````python
verified_orders = orders.loc[
    orders["record_status"].ne("ingestion_error")
].copy()

logical_orders = (
    verified_orders
    .drop_duplicates(subset="order_id", keep="first")
    .copy()
)

pd.Series({
    "logical_rows": len(logical_orders),
    "unique_order_id": logical_orders["order_id"].nunique(),
})
````

행 수와 고유 주문 ID 수가 모두 992인지 확인한 뒤에만 공식 분석을 시작합니다.

````python
gross_total = logical_orders["sales"].sum()
discount_total = (logical_orders["sales"] - logical_orders["final_sales"]).sum()
final_total = logical_orders["final_sales"].sum()

pd.Series({
    "gross_sales": int(gross_total),
    "discount_amount": int(discount_total),
    "final_sales": int(final_total),
})
````

### 첫 공식 매출 Evidence

- Gross Sales: **78,000,000원**
- Discount Amount: **4,000,000원**
- Final Sales: **74,000,000원**

이 숫자들은 `logical_orders`가 992개의 고유 주문으로 확정된 뒤 처음 사용합니다.
