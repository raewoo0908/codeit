# Mission 4 — 조건에 맞는 상품 찾기

완성된 코드를 떠올리기 전에, 어떤 열에 어떤 조건을 만들어야 하는지 먼저 생각하세요.

```python
# 각 상품의 가격이 30,000원 이상인지 True/False로 확인하세요.
mask = products["price"] >= 30000
```

결과는 왜 상품 수와 같은 개수의 True/False로 나왔나요?

```python
# 가격이 30,000원 이상인 상품만 선택하세요.
products[mask]
```

```python
# 평점이 4.5 이상인 상품만 찾으세요.
mask = products["rating"] >= 4.5
products[mask]
```

```python
# 평점이 4.0 이상이면서 가격이 30,000원 이하인 상품을 찾으세요.
mask = (products["rating"] >= 4.0) & (products["price"] <= 30000)

products[mask]
```

```python
# 전자기기 또는 생활 카테고리 상품만 찾으세요.
mask = (products["category"] == "전자기기") | (products["category"] == "생활")

products[mask]
```

<details>
<summary>힌트 보기</summary>

각 조건을 따로 만든 뒤, 동시에 만족하면 `&`, 둘 중 하나면 `|`로 연결합니다. 각 조건은 괄호로 묶으세요.

</details>
