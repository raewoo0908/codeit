## 4. 행과 열을 함께 선택하기

`loc`은 **라벨**, `iloc`은 **위치**를 기준으로 찾습니다.

### loc: Location
- 이름 기준: 로우 이름이나 컬럼 이름(라벨)을 직접 적어서 데이터를 찾는다. 
    - '이름(name)' 또는 '라벨(label)'이란, '인덱스(index)'와 '컬럼(column)'을 통틀어 부르는 말이다.
        - 인덱스: 표의 행(row)를 식별하는 눈금이다. 0행은 Index 0, 1행은 index 1, 2행을 index 2 ....
        - 컬럼: 표의 열(column)을 식별하는 눈금이다.
        - 라벨 | 이름: 인덱스와 컬럼 자리에 실제로 적혀있는 문자나 숫자 **값**. pandas에서는 **라벨**과 **이름**을 같은 의미로 쓴다. 
            - 행 라벨(Row label): 인덱스 자리에 적힌 값들이다. students 데이터프레임에서는 0, 1, 2, 3, 4, 5
            - 열 라벨(Column label): 컬럼 자리에 적힌 값들이다. students 데이터프레임에서는 "name", "score", "city", "study_hours"
    - ex) loc[1]: 1번 행을 리턴한다.
    - ex)
- 범위 특징: 시작점부터 끝점까지 이름 그대로 모두 포함한다. 
    - ex) loc[0:2]: 0, 1, 2번 행을 모두 리턴한다.

```python
students
```

```python
students.loc[0]
students.loc[1] # 각 컬럼의 1번 인덱스를 리턴한다. 
```

```python
type(students.loc[1]) # 타입은 Series다. 
```

```python
students.loc[0:3]
```

```python
# # 행 인덱스(라벨)를 '민수', '지수', '영희'로 바꾸었다고 가정해봅시다.
students_new = students.set_index("name")

# 이제 loc을 쓸 때 문자로 된 '이름(라벨)'을 넣어야 합니다.
# 아래 코드는 "민수" 행부터 "영희" 행까지 "score", "city" 컬럼에 있는 값들을 DataFrame으로 리턴하라는 의미이다.
students_new.loc["민수":"영희", ["score", "city"]]

# 행 인덱스에 붙어있는 '라벨'('이름')이 사람 이름이 되었다.
```

```python
# 행을 필터링 할 수도 있다. 
# 80점 이상인 학생들만 모든 컬럼에 대해서 받아보고 싶다면? -> DataFrame(여러 컬럼 보니깐)
students.loc[students["score"] >= 80]

# 80점 이상인 학생들만 "name" 컬럼에 대해서 받아보고 싶다면? -> Series(컬럼 하나밖에 안 보니깐)
students.loc[students["score"] >= 80, "name"]
```

### iloc: Integer Location

- 0부터 시작하는 몇 번째 순서인 지 숫자로만 데이터를 찾는다.  
- `students.iloc[0]`은 0부터 시작해서 0 번째 순서인 행을 반환한다. 이 때 타입은 `Series`.
- `students.iloc[0:3]`은 0부터 시작해서 3개의 행을 반환한다. 즉 0,1,2번째 행을 반환한다. 이 때 타입은 `DataFrame`.

```python
students.iloc[0] # 0부터 시작해서 0번째 행을 반환한다.  

students.iloc[1] # 0부터 시작해서 1번째 행을 반환한다. 
```

```python
students.iloc[0:3] # 마지막 인덱스는 포함되지 않는다.
```

```python
type(students.iloc[1]) # 하나의 행만 리턴하면 Series다.
```

```python
type(students.iloc[0:3]) # 여러 개의 행을 리턴하면 DataFrame이다.
```

```python
# 컬럼을 지정할 수도 있다. 단, 인덱스로 지정해야 한다. loc(<행라벨 이름(들)>, <컬럼 라벨(들)>)에서는 '컬럼 라벨'로 보고싶은 컬럼을 지정할 수 있었던 것과 다르다.

students.iloc[0:2, 0:3]
```

`loc[0:2]`는 끝 라벨 2를 포함합니다. 몇 행과 몇 열이 나올지 예상해보세요.

```python
# 0부터 2개의 행, 즉 0행,1행을 리턴하는데, "name", "score" 컬럼만 반환하라.

students.loc[0:2, ["name", "score"]]
```

`iloc`의 위치 범위에서는 끝 위치가 포함되지 않습니다.

```python
students.iloc[0:3, 0:2]
```

```python
# 첫 번째 학생의 정보를 loc으로 확인하세요.
# 처음 3명의 name과 score를 loc으로 선택하세요.
# iloc으로 첫 2행, 첫 2열을 선택하세요.

students.loc[0]

students.loc[0:2, ["name", "score"]]

students.iloc[0:2, 0:2]
```

# 5. 조건에 맞는 데이터 찾기

80점 이상인 학생을 바로 고르기 전에, 각 학생이 조건에 맞는지 먼저 확인합니다.

실행하면 숫자, 이름, True/False 중 무엇이 나올까요?

```python
students["score"] >= 80
```

### Boolean Mask

조건을 적용하면 각 행에 대해 `True / False`가 만들어집니다.

`조건 → 각 행 비교 → True / False`

```python
mask = students["score"] >= 80
mask
```

이제 `True`인 행만 골라낼 수 있을까요?

```python
students
```

```python
students[mask]
```

```python
students[students["score"] >= 80]
```

숫자 비교와 문자열 비교도 `조건 → True/False → 선택`이라는 흐름은 같습니다.

```python
students["city"] == "서울"
```

```python
students[students["city"] == "서울"]
```

서울 학생이면서 점수가 80점 이상인 학생을 찾으려면 어떤 두 조건이 필요할까요?

```python
students[
    (students["city"] == "서울")
    & (students["score"] >= 80)
]
```

`&`는 그리고, `|`는 또는입니다. 여러 조건은 각각 괄호로 묶습니다.

```python
students[
    (students["city"] == "서울")
    | (students["city"] == "부산")
]
```

### Filtering 연습

한 문제씩 조건을 만들고, 그 조건이 `True`인 행을 선택하세요.

```python
# 점수가 85점 이상인 학생을 찾으세요.
mask = students["score"] >= 85

students[mask]
```

```python
# 부산 학생만 찾으세요.
mask = students["city"] == "부산"

students[mask]
```

```python
# 부산 학생이면서 점수가 85점 이상인 학생을 찾으세요.
mask = (students["city"] == "부산") & (students["score"] >= 85)

students[mask]
```

```python
# 서울 또는 부산에 사는 학생을 찾으세요.
mask = (students["city"] == "부산") | (students["city"] == "서울")

students[mask]
```

# 7. 작은 판매 데이터

```python
sales["sales"] = sales["price"] * sales["quantity"]
sales
```

```python
sales["discount"] = [0.1, 0.0, 0.2]
sales
```

```python
# 할인 전 sales와 discount를 이용해 discount_amount 열을 만들어보세요.
sales["discount_amount"] = sales["sales"] * sales["discount"]
sales
```
