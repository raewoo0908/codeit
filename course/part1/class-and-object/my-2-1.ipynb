# 실험버전 3

의문: 클래스 이름만 선언되어있고, 인스턴스 이름이 선언되어있지 않았을 때, 인스턴스가 클래스 이름에 접근하면, 그 이름의 생명주기는 어떻게 될까? 언제 클래스 이름을 본인 네임스페이스에 복사하고, 언제 그게 사라질까? 

```python
class Student:
    total_created = 0

    def __init__(self, name, score):
        self.name = name
        self.score = score
        Student.total_created += 1

    def increment_instance_count(self):
        print(f"BEFORE: name: {self.name}, total_created: {Student.total_created}, instance_total_created: {self.total_created}")
        self.total_created += 1
        print(f"AFTER: name: {self.name}, total_created: {Student.total_created}, instance_total_created: {self.total_created}")
        print("--------------------------------")

    def update_score(self, score):
        self.score = score 


try:
    student1 = Student('민수', 80)
    student2 = Student('지수', 90)

    # 민수.total_created 증가 -> 클래스 이름을 민수의 네임스페이스에 복사 : 2 -> 3으로 증가. 
    student1.increment_instance_count()

    # 철수 생성 -> Student.total_created 증가 : 2 -> 3
    student3 = Student('철수', 80)

    # 민수.total_created 증가 -> Student.total_created는 현재 3인데, 과연 여기서 민수.total_created는 4가 될까 5가 될까?
    student1.increment_instance_count()

    # 지수.total_created 증가 -> Student.total_created을 지수의 네임스페이스에 복사 : 3 -> 4로 증가.
    # 인스턴스에서 클래스 이름을 참조하는 시점에 클래스 이름 값을 가져온다는 것을 알 수 있다. 
    student2.increment_instance_count()

except (AssertionError, NotImplementedError, TypeError, ValueError) as error:
    print('미완료')
    print('확인할 내용:', error)
else:
    print('테스트 통과')
```
