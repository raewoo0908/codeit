# Chapter 03 세 가지 메서드 구분하기
필요한 정보가 객체인지, 클래스인지, 둘 다 아닌지 먼저 판단하세요.


```python
class Product:
    tax_rate = 0.1

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_final_price(self):
        # TODO 1: 객체 가격과 클래스 세율을 사용해 최종 가격을 반환하세요.
        return self.price * (1 + Product.tax_rate)

    @classmethod
    def set_tax_rate(cls, rate):
        # TODO 2: 0 이상인 세율만 클래스 변수에 저장하세요.
        # 잘못된 값에는 ValueError를 발생시키세요.
        if rate < 0:
            raise ValueError('세율은 0 이상이어야 합니다.')
        cls.tax_rate = rate

    @staticmethod
    def is_valid_name(name):
        # TODO 3: 공백이 아닌 문자열이면 True를 반환하세요.
        if isinstance(name, str) and name.strip() != '':
            return True
        return False


try:
    keyboard = Product('Keyboard', 10000)
    mouse = Product('Mouse', 20000)

    assert keyboard.get_final_price() == 11000
    assert mouse.get_final_price() == 22000

    Product.set_tax_rate(0.2)
    assert Product.tax_rate == 0.2
    assert keyboard.get_final_price() == 12000
    assert mouse.get_final_price() == 24000

    assert Product.is_valid_name('Keyboard') is True
    assert Product.is_valid_name('   ') is False
    assert Product.is_valid_name(100) is False

    try:
        Product.set_tax_rate(-0.1)
    except ValueError:
        pass
    else:
        raise AssertionError('음수 세율에는 ValueError가 필요합니다.')

except (AssertionError, NotImplementedError, TypeError, ValueError) as error:
    print('미완료')
    print('확인할 내용:', error)
else:
    print('테스트 통과')
```

# 강사님 정답

```python
class Product:
    tax_rate = 0.1

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_final_price(self):
        # TODO 1: 객체 가격과 클래스 세율을 사용해 최종 가격을 반환하세요.
        return round(self.price * (1 + self.tax_rate))

    @classmethod
    def set_tax_rate(cls, rate):
        # TODO 2: 0 이상인 세율만 클래스 변수에 저장하세요.
        # 잘못된 값에는 ValueError를 발생시키세요.
        if rate < 0:
            raise ValueError('세율은 0 이상이어야 합니다.')
        cls.tax_rate = rate

    @staticmethod
    def is_valid_name(name):
        # TODO 3: 공백이 아닌 문자열이면 True를 반환하세요.
        return isinstance(name, str) and bool(name.strip())


try:
    keyboard = Product('Keyboard', 10000)
    mouse = Product('Mouse', 20000)

    assert keyboard.get_final_price() == 11000
    assert mouse.get_final_price() == 22000

    Product.set_tax_rate(0.2)
    assert Product.tax_rate == 0.2
    assert keyboard.get_final_price() == 12000
    assert mouse.get_final_price() == 24000

    assert Product.is_valid_name('Keyboard') is True
    assert Product.is_valid_name('   ') is False
    assert Product.is_valid_name(100) is False

    try:
        Product.set_tax_rate(-0.1)
    except ValueError:
        pass
    else:
        raise AssertionError('음수 세율에는 ValueError가 필요합니다.')

except (AssertionError, NotImplementedError, TypeError, ValueError) as error:
    print('미완료')
    print('확인할 내용:', error)
else:
    print('테스트 통과')
```
