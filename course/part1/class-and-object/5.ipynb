학습 목표

2. 객체에 * 연산 규칙을 정의할 수 있다.
3. bool()이 __bool__()을 호출한다는 것을 설명할 수 있다.
4. 연산 결과로 새로운 객체를 반환할 수 있다.


```python
class AppleBox:
    """사과를 담는 상자 클래스."""

    def __init__(self, apple_count=0):
        if not isinstance(apple_count, int):
            raise TypeError('사과 개수는 정수여야 합니다.')

        if apple_count < 0:
            raise ValueError('사과 개수는 0 이상이어야 합니다.')

        self.apple_count = apple_count

    def __str__(self):
        return f'사과 {self.apple_count}개'

    def __repr__(self):
        return f'AppleBox(apple_count={self.apple_count})'

    def __add__(self, other):
        """두 상자를 합친 새로운 상자를 반환한다."""

        if not isinstance(other, AppleBox):
            return NotImplemented

        new_count = self.apple_count + other.apple_count

        return AppleBox(new_count)

    def __mul__(self, number):
        """사과 개수를 배수만큼 늘린 새 상자를 반환한다."""

        if not isinstance(number, int):
            return NotImplemented

        if number < 0:
            raise ValueError('곱하는 수는 0 이상이어야 합니다.')

        new_count = self.apple_count * number

        return AppleBox(new_count)

    def __bool__(self):
        """사과가 있으면 True, 빈 상자이면 False를 반환한다."""

        return self.apple_count > 0
```

## 1. 객체 생성하기


```python
box1 = AppleBox(5)
box2 = AppleBox(3)
empty_box = AppleBox()

print(box1)
print(box2)
print(empty_box)
```

## 2. 두 상자 더하기


### 예측과 판단

실행하기 전에 결과를 예측하세요.


```python
combined_box = box1 + box2

print()
print('box1:', box1)
print('box2:', box2)
print('합친 상자:', combined_box)
```

box1 + box2
      ↓
box1.__add__(box2)
      ↓

## 3. 원래 객체 확인하기


```python
print()
print('연산 후 box1:', box1)
print('연산 후 box2:', box2)
print('새로운 객체인가?', combined_box is not box1)
```

box1과 box2는 변경되지 않았다.

## 4. 상자 곱하기

## 5. 빈 상자 확인하기


```python
print()
print('box1:', bool(box1))
print('empty_box:', bool(empty_box))
```

bool(box1)
     ↓
box1.__bool__()

사과가 한 개 이상이면 True,
사과가 없으면 False를 반환한다.


## 6. if문에서 객체 사용하기


```python
print()

if box1:
    print('box1에는 사과가 있습니다.')
else:
    print('box1은 비어 있습니다.')

if empty_box:
    print('empty_box에는 사과가 있습니다.')
else:
    print('empty_box는 비어 있습니다.')
```

## 7. 지원하지 않는 자료형과 더하기


```python
print()

try:
    box1 + 5
except TypeError as error:
    print('연산 실패:', error)
```

__add__()가 직접 처리할 수 없는 상대를 만나면
예외를 바로 발생시키지 않고 NotImplemented를 반환한다.
파이썬은 다른 연산 방법도 찾은 뒤, 방법이 없으면
최종적으로 TypeError를 발생시킨다.


## 8. 잘못된 값으로 곱하기


```python
print()

try:
    box1 * -2
except ValueError as error:
    print('연산 실패:', error)
```

음수 배수는 AppleBox가 정한 규칙에 맞지 않으므로
__mul__()이 직접 ValueError를 발생시킨다.
