```python
years = np.arange(2010, 2020)
gdp = np.array([6610, 7637, 8127, 8885, 10385,
                12565, 14403, 12398, 8282, 10672])
```

```python
over_10000 = gdp > 10000

print("GDP:", gdp)
print("조건:", over_10000)
print("조건 dtype:", over_10000.dtype)
print("같은 shape:", gdp.shape == over_10000.shape)
```

```python
mask = gdp > 10000


print("8,000 이상 10,000 이하:", gdp >= 8000 and gdp <= 10000)
print("8,000 미만 또는 12,000 초과:", (gdp < 8000) + (gdp > 12000))
```

```python
# TODO: numpy array에서 논리연산자, 산술연산자의 구현과 용례!

print("8000 이상: ", gdp >= 8000)
print("10000 이하: ", gdp <= 10000)
print("8000 이상 AND 10,000 이하: ", (gdp >= 8000) * (gdp <= 10000)) # *
print("8000 이상 AND 10,000 이하: ", (gdp >= 8000) & (gdp <= 10000)) # AND

or_more_than_8000 = gdp[gdp >= 8000]
or_less_than_10000 = gdp[gdp <= 10000]
or_more_than_8000_and_or_less_than_10000 = gdp[(gdp >= 8000) * (gdp <= 10000)]
print("8000 이상 AND 10,000 이하(masked): ", or_more_than_8000_and_or_less_than_10000)
print("------------------------------------------------------------------------")
print("8000 미만: ", gdp < 8000)
print("10000 초과: ", gdp > 10000)
print("8000 미만 OR 12,000 초과: ", (gdp < 8000) + (gdp > 12000)) # +
print("8000 미만 OR 12,000 초과: ", (gdp < 8000) | (gdp > 12000)) # OR

less_than_8000_or_over_than_120000 = gdp[(gdp < 8000) + (gdp > 12000)]
print("8000 미만 OR 12,000 초과(masked): ", less_than_8000_or_over_than_120000)
```

1부터 100까지의 정수에서 짝수만 선택합니다.

설계 질문:

1. 1부터 100까지 만들 `arange`의 stop은? -> 101
2. 짝수는 2로 나눈 나머지가 무엇인가?       -> 0
3. 조건 결과를 배열 대괄호 어디에 넣는가?     -> 안에

```python
numbers = np.arange(1,101)

mask = numbers % 2 == 0

odds = numbers[mask]

print("numbers:\n", numbers.reshape(10,10))
print("odd numbers:\n", odds.reshape(5,10))
```
