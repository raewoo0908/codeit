- `cups` : 메뉴 5개의 하루 판매 잔 수 (1차원)

매장 4곳의 실내 온도가 실수로 기록되어 있습니다.

```python
temps = np.array([21.7, 24.3, 19.8, 26.5])
```

이 값을 정수형으로 바꾼 배열을 만들고, 변환 전/후의 값과 dtype을 출력하세요.
그리고 **원본 `temps`가 그대로 남아 있는지**도 함께 출력해 확인하세요.

변환 결과를 보고 '반올림인지 버림인지' 판단해 적으세요.

```text

```

````python
import numpy as np

temps = np.array([21.7, 24.3, 19.8, 26.5])


# ===== 여기에 코드를 작성하세요 =====

temps_copy = temps.astype(int) # temps의 복사본을 만들어서 int로 타입을 바꾼 다음 저장한다. 

print("BEFORE: ", temps)
print("BEFORE: ", temps.dtype) # 원본은 타입이 안 바뀌어있다. 그대로다. 
print("--------------------------------------------")
print("AFTER: ", temps_copy)   # 소수점은 버림되어 int로 변환된다.
print("AFTER: ", temps_copy.astype(int).dtype)
````
