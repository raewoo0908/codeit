````python
# [2/3] 라이브러리 불러오기 + 폰트 등록
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import seaborn as sns
import pandas as pd
import numpy as np

# 설치한 폰트 파일의 위치를 matplotlib에 알려준다 (이 줄이 없으면 한글이 깨집니다)
fm.fontManager.addfont('/usr/share/fonts/truetype/nanum/NanumGothic.ttf')

# seaborn 스타일을 쓰면서 폰트도 같이 지정한다
sns.set_theme(style="whitegrid",
              font="NanumGothic",
              rc={"axes.unicode_minus": False})

print("폰트:", plt.rcParams["font.family"])
````

````python
# [3/3] 데이터 준비 (강의 노트북과 같은 데이터입니다)

# --- week : 강남점 7일치 요약 ---
week = pd.DataFrame({
    "요일": ["월", "화", "수", "목", "금", "토", "일"],
    "매출": [120, 135, 128, 150, 210, 320, 290],   # 만원
    "방문객": [80, 88, 84, 96, 130, 190, 175],      # 명
})

# --- menu : 메뉴별 판매량 ---
menu = pd.DataFrame({
    "메뉴": ["아메리카노", "라떼", "콜드브루", "마카롱", "샌드위치"],
    "판매량": [420, 260, 180, 95, 130],
})

# --- sales : 3개 지점 x 28일 = 84일치 ---
rng = np.random.default_rng(0)
지점 = np.repeat(["강남", "홍대", "판교"], 28)
요일 = np.tile(["월", "화", "수", "목", "금", "토", "일"], 12)
기본방문 = np.repeat([120, 90, 70], 28)
주말배수 = np.where(np.isin(요일, ["토", "일"]), 1.6, 1.0)
방문객 = (기본방문 * 주말배수 + rng.normal(0, 10, 84)).round()

sales = pd.DataFrame({"지점": 지점, "요일": 요일, "방문객": 방문객})
객단가 = np.where(np.isin(요일, ["토", "일"]), 2.0, 2.4)
sales["일매출"] = (sales["방문객"] * 객단가 + rng.normal(0, 20, 84)).round()

# 요일 순서는 계속 쓰므로 미리 만들어 둡니다
순서 = ["월", "화", "수", "목", "금", "토", "일"]

print(sales.shape)
print(sales.head())
````
